```python
import requests
import pandas as pd

URL = "https://data.cityofnewyork.us/resource/erm2-nwe9.csv"

params = {
    "$select": "date_trunc_ymd(created_date) AS day, count(*) AS requests",
    "$where": "created_date >= '2021-01-01T00:00:00' "
              "AND created_date < '2026-01-01T00:00:00'",
    "$group": "day",
    "$order": "day",
    "$limit": 5000,
}

response = requests.get(URL, params=params, timeout=120)
response.raise_for_status()

with open("data/daily_requests.csv", "wb") as f:
    f.write(response.content)

df = pd.read_csv("data/daily_requests.csv", parse_dates=["day"])
print(df.head())
print(df.shape)
```
