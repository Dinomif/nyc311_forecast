```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/daily_requests.csv", parse_dates=["day"])
s = df.set_index("day")["requests"]

def mae(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))

def mape(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100

train = s[:"2024-12-31"]
test = s["2025-01-01":]
print(len(train), len(test))
```

```python
as_yesterday = s.shift(1)[test.index]
as_week_ago = s.shift(7)[test.index]

for name, pred in [("Ошибки в подходе Как вчера", as_yesterday), ("Ошибка в подходе Как неделю назад", as_week_ago)]:
    print(f"{name}: MAE = {mae(test, pred):.0f}, MAPE = {mape(test, pred):.1f}%")
```
